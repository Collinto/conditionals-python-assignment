```python
name = input("Enter a name: ")

vowels = ["a", "e", "i", "o", "u"]
consonants=["b", "d", "f", "g", "h", "j", "k", "l", "m", "n", "p", "q","r","s", "t", "v", "x", "y", "z"]

if name[0] in vowels:
  print("First letter is vowel")
elif name[0] in consonants:
    print("First letter is consonant")
else:
  print("First character is not a letter")
```
