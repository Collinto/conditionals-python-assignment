```python
# Age Grading
```

```python
age = int(input("Enter Age: "))

match age:
  case a if a >= 0 and a <= 12:
    print("Child")
  case a if a >= 13 and a <= 19:
    print("Teenager")
  case a if a >= 20 and a <= 64:
    print("Adult")
  case a if a >= 65:
    print("Senior")
  case _:
    print("Not child")
```
