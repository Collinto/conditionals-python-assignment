```python
sentence= input("Enter a sentence: ")
fl=sentence[0].upper()
mark=[".", "!", "?"]

if sentence[0] == fl and sentence[-1] in mark:
  print("Proper Sentence")
else:
  print("Improper Sentence ")
```
