```python
pd = input("Enter a password of atleast 8 characters: ")
er = "Invalid Password"

if len(pd) < 8:
  print(er)
elif pd == "password":
  print(er)
elif pd == "123456":
  print(er)
else:
  print("Valid Password")
```
