```python
word = input("Enter a word: ")

alt = list(word)
alt.reverse()
alts = "".join(alt)

print(f"{word} was flipped to: {alts}")

if alts == word:
  print("It's a Palindrome")
else:
  print("It's not a Palindrome")
```
