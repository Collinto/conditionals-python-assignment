```python
phone = input("Enter phone number: ")
err = "Invalid Phone Number"


if len(phone) != 12:
  print(err)
elif phone[3] != "-" and phone[7] != "-":
  print(err)
else:
  number=phone[0:3] + phone[4:7] + phone[8:12]
  if number.isdigit:
    print("Valid number")
  else:
    print(err)

```
